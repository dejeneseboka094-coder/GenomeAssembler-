```python
# ============================================================
# QUESTION 3 — NAIVE PATTERN MATCHING
# ============================================================

T = "ACGTACGTACGT"
P = "TAC"

print("=" * 60)
print("QUESTION 3 — NAIVE PATTERN MATCHING")
print("=" * 60)

print("Text    :", T)
print("Pattern :", P)

n = len(T)
m = len(P)

occurrences = []

print("\nAlignments:")
print("-" * 60)

# Check every possible alignment
for i in range(n - m + 1):

    print("\nAlignment at position", i + 1)

    # Show the text section being compared
    text_part = T[i:i + m]

    print("Text section :", text_part)
    print("Pattern      :", P)

    match = True

    # Compare characters one by one
    for j in range(m):

        text_char = T[i + j]
        pattern_char = P[j]

        if text_char == pattern_char:
            print(
                "Character", j + 1,
                ":", text_char,
                "=", pattern_char,
                "MATCH"
            )
        else:
            print(
                "Character", j + 1,
                ":", text_char,
                "!=",
                pattern_char,
                "MISMATCH"
            )
            match = False

    # Check whether the entire pattern matched
    if match:
        print("Result: COMPLETE MATCH")
        occurrences.append(i + 1)
    else:
        print("Result: MISMATCH")


# Final result
print("\n" + "=" * 60)
print("FINAL RESULT")
print("=" * 60)

print("Pattern:", P)
print("Text   :", T)

print("Pattern occurs at positions:", occurrences)
```

```python
# ============================================================
# QUESTION 4 — LPS ARRAY
# ============================================================

P = "ABABCABAB"


def compute_lps(pattern):
    """
    Compute the LPS (Longest Proper Prefix which is also Suffix)
    array for a given pattern.
    """

    m = len(pattern)

    # Create an LPS array filled with zeros
    lps = [0] * m

    length = 0
    i = 1

    while i < m:

        # If characters match
        if pattern[i] == pattern[length]:

            length += 1
            lps[i] = length
            i += 1

        # If characters do not match
        else:

            if length != 0:
                length = lps[length - 1]

            else:
                lps[i] = 0
                i += 1

    return lps


# Calculate LPS
lps = compute_lps(P)


# ------------------------------------------------------------
# Display the result
# ------------------------------------------------------------

print("=" * 50)
print("QUESTION 4 — LPS ARRAY")
print("=" * 50)

print("Pattern:", P)

print("\nIndex   Character   LPS")
print("-" * 30)

for i in range(len(P)):
    print(f"{i:<8}{P[i]:<12}{lps[i]}")

print("\nLPS Array:")
print(lps)
```

```python
# ============================================================
# QUESTION 5 — KMP PATTERN MATCHING
# ============================================================

text = "ABABDABABCABAB"
pattern = "ABABCABAB"

print("=" * 60)
print("QUESTION 5 — KMP PATTERN MATCHING")
print("=" * 60)

print("Text    :", text)
print("Pattern :", pattern)


# ------------------------------------------------------------
# Step 1 — Calculate the LPS array
# ------------------------------------------------------------

lps = compute_lps(pattern)

print("\n1. LPS ARRAY")
print("-" * 40)

print("Pattern:", pattern)
print("LPS    :", lps)


# ------------------------------------------------------------
# Step 2 — KMP search
# ------------------------------------------------------------

def kmp_search(text, pattern):
    """
    KMP pattern matching algorithm.
    Returns the starting positions of the pattern.
    """

    n = len(text)
    m = len(pattern)

    lps = compute_lps(pattern)

    i = 0
    j = 0

    occurrences = []

    print("\n2. MAJOR MATCHING STEPS")
    print("-" * 60)

    while i < n:

        # Characters match
        if text[i] == pattern[j]:

            print(
                "MATCH:",
                text[i],
                "at text position",
                i + 1,
                "with pattern position",
                j + 1
            )

            i += 1
            j += 1

            # Complete pattern found
            if j == m:

                position = i - m + 1

                occurrences.append(position)

                print(
                    ">>> COMPLETE MATCH FOUND "
                    "starting at position",
                    position
                )

                # Continue searching
                j = lps[j - 1]

        # Characters do not match
        else:

            print(
                "MISMATCH:",
                text[i],
                "and",
                pattern[j],
                "at text position",
                i + 1
            )

            # Use LPS to avoid unnecessary comparisons
            if j != 0:

                print(
                    "Using LPS value:",
                    lps[j - 1]
                )

                j = lps[j - 1]

            else:

                i += 1

    return occurrences


# Run KMP
occurrences = kmp_search(text, pattern)


# ------------------------------------------------------------
# Step 3 — Final result
# ------------------------------------------------------------

print("\n3. PATTERN LOCATION")
print("-" * 40)

print("Pattern:", pattern)
print("Occurrences:", occurrences)

print("\nFinal Answer:")
print(
    "The pattern ABABCABAB occurs at position 6 "
    "using 1-based indexing."
)
```

```python
# ============================================================
# QUESTION 6 — SUFFIX ARRAYS
# ============================================================

S = "BANANA$"

print("=" * 60)
print("QUESTION 6 — SUFFIX ARRAYS")
print("=" * 60)

print("String:", S)


# ------------------------------------------------------------
# 6(a) List all suffixes with their starting positions
# ------------------------------------------------------------

print("\n6(a) ALL SUFFIXES")
print("-" * 40)

suffixes = []

for i in range(len(S)):
    suffix = S[i:]
    suffixes.append((i, suffix))
    print("Position", i, ":", suffix)


# ------------------------------------------------------------
# 6(b) Sort the suffixes lexicographically
# ------------------------------------------------------------

print("\n6(b) SORTED SUFFIXES")
print("-" * 40)

sorted_suffixes = sorted(suffixes, key=lambda x: x[1])

for rank, (position, suffix) in enumerate(sorted_suffixes):
    print(
        "Rank",
        rank,
        "| Position",
        position,
        "| Suffix:",
        suffix
    )


# ------------------------------------------------------------
# 6(c) Construct the suffix array
# ------------------------------------------------------------

print("\n6(c) SUFFIX ARRAY")
print("-" * 40)

suffix_array = []

for position, suffix in sorted_suffixes:
    suffix_array.append(position)

print("Suffix Array:", suffix_array)


# ------------------------------------------------------------
# 6(d) Search for ANA using the suffix array
# ------------------------------------------------------------

print("\n6(d) SEARCH FOR 'ANA'")
print("-" * 40)

pattern = "ANA"

print("Pattern:", pattern)

matches = []

for position in suffix_array:

    suffix = S[position:]

    print(
        "Checking position",
        position,
        ":",
        suffix
    )

    if suffix.startswith(pattern):
        matches.append(position)
        print("  --> MATCH")


print("\nPositions where ANA occurs:", matches)


# ------------------------------------------------------------
# Final answers
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL ANSWERS — QUESTION 6")
print("=" * 60)

print("""
6(a) All suffixes:

Position 0: BANANA$
Position 1: ANANA$
Position 2: NANA$
Position 3: ANA$
Position 4: NA$
Position 5: A$
Position 6: $

6(b) Sorted suffixes:

$        -> position 6
A$       -> position 5
ANA$     -> position 3
ANANA$   -> position 1
BANANA$  -> position 0
NA$      -> position 4
NANA$    -> position 2

6(c) Suffix Array:
[6, 5, 3, 1, 0, 4, 2]

6(d) ANA occurs at positions:
[1, 3]
""")
```

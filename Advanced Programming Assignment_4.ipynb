## Exercise 1 — Write a GC Content Function

Write a function called `calculate_gc()` that accepts a DNA sequence and returns its GC content as a percentage.

```python
def calculate_gc(sequence):
    sequence = sequence.upper()  # Convert to uppercase

    gc_count = sequence.count('G') + sequence.count('C')
    gc_percentage = (gc_count / len(sequence)) * 100

    return gc_percentage


dna = input("Enter a DNA sequence: ")
print("GC Content:", calculate_gc(dna), "%")
```

Why is GenBank more informative than FASTA for genome annotation analysis?

Discuss with examples such as:

- gene names
- CDS coordinates
- protein products
- organism metadata
- references

GenBank is more informative than FASTA for genome annotation because it includes both sequence data and rich biological annotations, while FASTA only stores raw sequences.

#### 1. Gene names
FASTA: Only sequence + simple header.
GenBank: Explicit gene annotations.
Example: /gene="BRCA1".
Enables direct identification of functional genes.
#### 2. CDS coordinates
FASTA: No positional information.
GenBank: Defines exact coding regions.
Example: CDS 100..900.
Essential for gene structure, exon mapping, and translation accuracy.
#### 3. Protein products
FASTA: No functional output information.
GenBank: Links genes to protein products.
Example: /product="DNA polymerase".
Supports functional genomics and pathway analysis.
#### 4. Organism metadata
FASTA: Minimal or no biological context.
GenBank: Includes full organism details.
Example: species, strain, isolate, taxonomy.
Important for comparative genomics and evolutionary studies.
#### 5. References
FASTA: No literature support.
GenBank: Includes citations and publications.
Example: author list, journal references.
Allows validation and traceability of annotations.

## Exercise 7 — Mutation Interpretation

Align the following sequences and identify the difference:

- Sequence 1: `ATGCCCTAG`
- Sequence 2: `ATGCGCTAG`

```python
from Bio import Align

# Get sequences from user
seq_a = input("Enter Sequence 1: ").upper()
seq_b = input("Enter Sequence 2: ").upper()

# Create aligner
aligner = Align.PairwiseAligner()

# Perform alignment
alignments = aligner.align(seq_a, seq_b)

# Best alignment
best_alignment = alignments[0]

print("\n===== ALIGNMENT RESULT =====\n")
print(best_alignment)

print("Best Alignment Score:", best_alignment.score)

# -----------------------------
# Mutation analysis
# -----------------------------

print("\n===== MUTATION ANALYSIS =====")

if len(seq_a) == len(seq_b):
    differences = []

    for i in range(len(seq_a)):
        if seq_a[i] != seq_b[i]:
            differences.append((i + 1, seq_a[i], seq_b[i]))

    if len(differences) == 0:
        print("No mutation detected (sequences are identical).")
    else:
        print(f"Total mutations: {len(differences)}")
        for pos, a, b in differences:
            print(f"Position {pos}: {a} → {b}")
        print("Type: Single nucleotide substitution (SNP)")
else:
    print("Sequences have different lengths → possible insertion/deletion mutation.")
```

## Exercise 10 — Improve the Pipeline

Modify the pipeline to also report:

- reverse complement
- number of stop codons in translated protein
- whether the sequence contains ambiguous bases such as N
- AT content
- sequence category: high GC, medium GC, or low GC

```python
from Bio import SeqIO
from Bio.Seq import Seq
import pandas as pd

# GC function (if not already defined)
def calculate_gc(seq):
    seq = str(seq).upper()
    return ((seq.count("G") + seq.count("C")) / len(seq)) * 100


def analyze_fasta(input_fasta, output_csv):
    results = []

    for record in SeqIO.parse(input_fasta, "fasta"):
        seq = str(record.seq).upper()
        seq_obj = Seq(seq)

        # Basic features
        length = len(seq)
        gc = calculate_gc(seq)
        at_content = ((seq.count("A") + seq.count("T")) / length) * 100

        # Reverse complement
        rev_comp = str(seq_obj.reverse_complement())

        # Translation (if valid coding sequence)
        protein = ""
        stop_count = 0

        if seq.startswith("ATG") and length % 3 == 0:
            protein = str(seq_obj.translate(to_stop=False))
            stop_count = protein.count("*")

        # Ambiguous bases check
        has_ambiguous = "N" in seq

        # GC category
        if gc >= 60:
            gc_category = "High GC"
        elif gc >= 40:
            gc_category = "Medium GC"
        else:
            gc_category = "Low GC"

        results.append({
            "id": record.id,
            "description": record.description,
            "length": length,
            "gc_content": round(gc, 2),
            "at_content": round(at_content, 2),
            "starts_with_ATG": seq.startswith("ATG"),
            "length_multiple_of_3": length % 3 == 0,
            "translated_sequence": protein,
            "stop_codon_count": stop_count,
            "reverse_complement": rev_comp,
            "contains_N": has_ambiguous,
            "gc_category": gc_category
        })

    df = pd.DataFrame(results)
    df.to_csv(output_csv, index=False)
    return df


# Run pipeline
pipeline_results = analyze_fasta("example_sequences.fasta", "sequence_summary.csv")
print(pipeline_results)
```

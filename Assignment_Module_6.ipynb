## Question 1

Define the following algorithmic design paradigms and provide one biological or bioinformatics example for each.

### (a) Brute Force



Brute Force is an algorithmic approach that systematically examines every possible solution until the correct one is found. Although it always produces the correct result, it is often computationally expensive because it performs an exhaustive search.

### Biological Example

A brute-force algorithm can be used to search for a DNA motif by comparing the motif with every possible position in a DNA sequence until all exact matches are identified.

### (b) Divide and Conquer



Divide and Conquer is an algorithmic paradigm that divides a large problem into smaller independent subproblems, solves each recursively, and combines the individual solutions to obtain the final result. This approach improves computational efficiency for many large-scale problems.

### Biological Example

Merge Sort is commonly used to sort millions of DNA sequencing reads according to genomic coordinates before downstream analyses such as genome assembly or variant detection.

### (c) Greedy Algorithm

A Greedy Algorithm constructs a solution by selecting the locally optimal choice at every step. It does not reconsider previous decisions and aims to produce a globally optimal solution with minimal computational cost.

### Biological Example

The Neighbor-Joining algorithm used for phylogenetic tree construction repeatedly joins the pair of taxa with the smallest evolutionary distance, following a greedy strategy.

### (d) Dynamic Programming

Dynamic Programming solves complex problems by dividing them into overlapping subproblems, storing intermediate solutions, and reusing them whenever needed. This significantly reduces redundant calculations and improves efficiency.

### Biological Example

The Needleman–Wunsch algorithm performs global DNA or protein sequence alignment using dynamic programming to identify the optimal alignment between two biological sequences.

### (e) Backtracking

Backtracking is a recursive search technique that incrementally constructs candidate solutions. If a partial solution cannot lead to a valid complete solution, the algorithm abandons that path and explores alternative possibilities.

### Biological Example

Backtracking can be used to identify a common DNA motif shared among multiple sequences by exploring candidate motifs and discarding those that fail to satisfy the required matching criteria.

## Question 4

Implement Merge Sort using the Divide and Conquer paradigm.

**Requirements**

- Input a random list of integers.
- Display the original array.
- Display the sorted array.
- Count the number of recursive calls.

````python

# Merge Sort using Divide and Conquer

import random

# Counter for recursive calls
recursive_calls = 0

def merge_sort(arr):
    global recursive_calls
    recursive_calls += 1

    if len(arr) <= 1:
        return arr

    # Divide
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Conquer (Merge)
    merged = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1

    merged.extend(left[i:])
    merged.extend(right[j:])

    return merged


# Generate a random list of integers
random.seed(42)
numbers = random.sample(range(1, 101), 15)

print("=" * 50)
print("Original Array")
print("=" * 50)
print(numbers)

sorted_numbers = merge_sort(numbers)

print("\n" + "=" * 50)
print("Sorted Array")
print("=" * 50)
print(sorted_numbers)

print("\n" + "=" * 50)
print("Number of Recursive Calls")
print("=" * 50)
print(recursive_calls)
````

## Explanation

Merge Sort is a classic example of the **Divide and Conquer** algorithmic paradigm. The algorithm recursively divides the original array into smaller subarrays until each subarray contains only one element. These subarrays are then merged in sorted order to produce the final sorted array.

For this implementation, a random list of 15 integers was generated. The program first displays the original unsorted array, then recursively sorts the elements using Merge Sort, and finally displays the sorted array. A global counter records the total number of recursive calls made during execution, providing an indication of the recursive nature of the algorithm.

Merge Sort has a worst-case and average-case time complexity of **O(n log n)**, making it one of the most efficient comparison-based sorting algorithms for large datasets. Unlike simpler algorithms such as Bubble Sort, Merge Sort maintains consistent performance regardless of the initial ordering of the input data.

## Question 5

Implement a simple DNA motif search using the Brute Force algorithm.

**Requirements**

- Search for the motif **"ATG"** in the DNA sequence.
- Print all starting positions where the motif occurs.
- Count the total number of comparisons performed.
- Explain why BLAST is much faster than the brute-force approach.

````python

# Brute Force DNA Motif Search


# DNA sequence
dna_sequence = "ATGCGATATGCCATGGATGATCGATG"

# Motif to search
motif = "ATG"

positions = []
comparisons = 0

# Brute-force search
for i in range(len(dna_sequence) - len(motif) + 1):

    match = True

    for j in range(len(motif)):
        comparisons += 1

        if dna_sequence[i + j] != motif[j]:
            match = False
            break

    if match:
        positions.append(i)

print("=" * 50)
print("DNA Sequence")
print("=" * 50)
print(dna_sequence)

print("\nMotif:", motif)

print("\n" + "=" * 50)
print("Motif Found at Positions")
print("=" * 50)
print(positions)

print("\n" + "=" * 50)
print("Total Character Comparisons")
print("=" * 50)
print(comparisons)
````

## Explanation

The brute-force motif search algorithm scans the DNA sequence one position at a time. At each position, it compares the motif ("ATG") with the corresponding characters in the DNA sequence. If all characters match, the starting position of the motif is recorded. The algorithm also counts the total number of character comparisons performed during the search.

For the given DNA sequence, the motif **ATG** occurs at several positions, demonstrating how the algorithm systematically examines every possible alignment of the motif. While this exhaustive approach guarantees that all exact matches are found, it becomes computationally expensive for long genomic sequences because every potential starting position must be examined.

### Why is BLAST much faster than the Brute Force algorithm?

BLAST (Basic Local Alignment Search Tool) is much faster because it does **not** compare the query sequence against every possible position in the database. Instead, BLAST first identifies short matching words (called *seeds*) between the query and database sequences. It then extends only these promising matches to produce alignments. This heuristic strategy greatly reduces the number of comparisons, allowing BLAST to search very large genomic and protein databases within seconds while maintaining high sensitivity. In contrast, the brute-force algorithm performs an exhaustive search, making it much slower for large-scale bioinformatics applications.

### (b) Explain how pruning reduces computational time.

Pruning is a technique used in backtracking algorithms to eliminate branches of the search tree that cannot produce a valid solution. Instead of exploring every possible combination, the algorithm evaluates partial solutions as they are constructed. If a partial solution violates the required constraints or cannot lead to a successful outcome, that branch is immediately abandoned.

In DNA motif discovery, for example, if the first few nucleotides of a candidate motif do not match the required pattern across the sequences, the algorithm stops exploring that candidate and backtracks to evaluate another possibility. By avoiding unnecessary computations, pruning substantially decreases the search space and improves execution time, especially when analyzing large biological datasets.

### (c) Compare Backtracking with Brute Force.

| Feature | Backtracking | Brute Force |
|----------|--------------|-------------|
| Search Strategy | Explores candidate solutions and abandons invalid branches early | Examines every possible solution without elimination |
| Pruning | Uses pruning to reduce the search space | No pruning; every possibility is evaluated |
| Computational Efficiency | More efficient because unnecessary searches are avoided | Less efficient due to exhaustive search |
| Memory Usage | Moderate because of recursive calls | Usually low but requires more computation |
| Bioinformatics Example | DNA motif discovery with constraint checking | Exact DNA motif search by checking every position |
| Suitable For | Constraint satisfaction and combinatorial optimization problems | Small datasets or exhaustive searches requiring guaranteed completeness |

### Summary

Backtracking is an efficient search strategy for solving combinatorial problems in bioinformatics because it avoids exploring branches that cannot produce valid solutions. Through pruning, it significantly reduces computational time compared with the brute-force approach while still guaranteeing that all valid solutions are found. Consequently, backtracking is widely used in applications such as DNA motif discovery, sequence analysis, and other biological optimization problems where exhaustive search would be computationally expensive.
